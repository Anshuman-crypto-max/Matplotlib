### Real World Example

```python
gas=pd.read_csv('gas_prices.csv')
plt.figure(figsize=[8,6])
# plt.plot(gas.Year,gas.USA,'r.-',label='USA')
# plt.plot(gas.Year,gas.Canada,'b.-',label='Canada')
# plt.plot(gas['Year'],gas['South Korea'],'g.-',label='South Korea')
# plt.plot(gas.Year,gas.Australia,'y.-',label='Australia')
# plt.title('USA V/S Canada V/S South Korea Gas Prices')
for country in gas:
    if country!='Year':
        plt.plot(gas.Year,gas[country],marker='.',label=country)
plt.ylabel('US Dollar V/S Gallon')
plt.xticks(gas.Year[::3].tolist()+[2011])
plt.savefig('Gas_Price Across World(Updated).png',dpi=300)
plt.legend()
plt.show()
```

### Load Fifa Data

```python
fifa=pd.read_csv('fifa_eda.csv')
fifa.head(5)
```

### Histogram

```python
bins=[40,50,60,70,80,90,100]

plt.hist(fifa.Overall,bins=bins,color='#75a832')
plt.xticks(bins)
plt.ylabel('Number of Players')
plt.xlabel('Skill Level')
plt.title('Distribution of Player Skills in FIFA 2019')
plt.savefig('Distribution  of Players skills in FIFA 2019.png',dpi=300)
plt.show()
```

### Pie Charts

```python
left=fifa.loc[fifa['Preferred Foot']=='Left'].count()[0]
right=fifa.loc[fifa['Preferred Foot']=='Right'].count()[0]
```

```python
labels=['Left','Right']
plt.pie([left,right],labels=labels,autopct='%.2f %%')
plt.title('Foot Preference of FIFA Players')
plt.savefig('Foot Prefference of FIFA Players.png',dpi=300)
plt.show()
```

```python
light=fifa.loc[fifa.Weight<125].count()[0]
light_medium=fifa.loc[(fifa.Weight>=125) &(fifa.Weight<150)].count()[0]
medium=fifa.loc[(fifa.Weight>=150) & (fifa.Weight<175)].count()[0]
heavy_medium=fifa.loc[(fifa.Weight>=175) & (fifa.Weight<200)].count()[0]
heavy=fifa.loc[fifa.Weight>=200].count()[0]
heavy_medium
```

```python
weights=[light,light_medium,medium,heavy_medium,heavy]
labels=['Under 125','125-150','150-175','175-200','200+']
explode=[1,.1,0,0.1,0.2]
plt.pie(weights,labels=labels,autopct='%.2f %%',explode=explode)
plt.title('Weight Distribution of FIFA Players(in lbs)')
plt.savefig('Weight Distribution of FIFA Pla.png',dpi=300)
plt.show()
```

```python
barcelona=fifa.loc[fifa.Club=='FC Barcelona']['Overall']
madrid=fifa.loc[fifa.Club=='Real Madrid']['Overall']
ner=fifa.loc[fifa.Club=='New England Revolution']['Overall']
plt.figure(figsize=[5,8])
labels=['FC Barcelona','Real Madrid','NE Revolution']
plt.boxplot([barcelona,madrid,ner],labels=labels)
plt.title('Professional Team Comparison')
plt.ylabel('FIFA Overall Ratings')
# plt.xlabel('FIFA Teams')
plt.show()
```
